```python
# Import bibliotek
import pandas as pd
import matplotlib.pyplot as plt

# Przykładowe dane
data = {
    'Miesiac': ['Styczen', 'Luty', 'Marzec', 'Kwiecien'],
    'Uzytkownicy': [120, 150, 210, 300]
}

# Tworzenie DataFrame
df = pd.DataFrame(data)

# Wyświetlenie tabeli
print(df)

# Wykres
plt.plot(df['Miesiac'], df['Uzytkownicy'], marker='o')
plt.title('Liczba użytkowników')
plt.xlabel('Miesiąc')
plt.ylabel('Użytkownicy')
plt.grid(True)

plt.show()
```
